# Two Particles in Circular Motion and Their Center of Mass

## Original problem

Two particles of masses $m_1$ and $m_2$ move uniformly in different circles of radii $R_1$ and $R_2$ about the origin in the $xy$-plane. The coordinates of particle 1 and of the center of mass are

$$x_1(t)=2\cos(4t), \qquad y_1(t)=2\sin(4t),$$

$$x_{\mathrm{CM}}(t)=8\cos(4t), \qquad y_{\mathrm{CM}}(t)=8\sin(4t),$$

where lengths are in meters and $t$ is in seconds.

1. Find the radius of the circle in which particle 1 moves.
2. Find $x_2(t)$, $y_2(t)$, and the radius $R_2$ of the circle in which particle 2 moves.

## Definition of center of mass

For particles with masses $m_i$ at position vectors $\mathbf r_i$, the center-of-mass position is the mass-weighted average

$$\boxed{\displaystyle \mathbf r_{\mathrm{CM}}=\frac{\sum_i m_i\mathbf r_i}{\sum_i m_i}}.$$

For two particles, this becomes

$$\boxed{\displaystyle \mathbf r_{\mathrm{CM}}=\frac{m_1\mathbf r_1+m_2\mathbf r_2}{m_1+m_2}}.$$

Equivalently, the definition applies component by component:

$$x_{\mathrm{CM}}=\frac{m_1x_1+m_2x_2}{m_1+m_2}, \qquad y_{\mathrm{CM}}=\frac{m_1y_1+m_2y_2}{m_1+m_2}.$$

## Analytical solution

### (a) Radius of particle 1

The distance from the origin is

$$R_1=\sqrt{x_1^2+y_1^2} = \sqrt{4\cos^2(4t)+4\sin^2(4t)}=2.$$

Therefore,

$$\boxed{R_1=2\ \mathrm{m}}.$$

### (b) Coordinates and radius of particle 2

Solving the center-of-mass definition for $\mathbf r_2$ gives

$$\mathbf r_2=\frac{(m_1+m_2)\mathbf r_{\mathrm{CM}}-m_1\mathbf r_1}{m_2}.$$

For the $x$-coordinate,

$$\begin{aligned}
x_2(t)&=\frac{(m_1+m_2)8\cos(4t)-m_1[2\cos(4t)]}{m_2}\\
&=\frac{6m_1+8m_2}{m_2}\cos(4t).
\end{aligned}$$

Similarly,

$$y_2(t)=\frac{6m_1+8m_2}{m_2}\sin(4t).$$

Thus particle 2 also moves counterclockwise with angular speed $4\ \mathrm{rad/s}$. Its radius is the common amplitude of its coordinates:

$$\boxed{\displaystyle R_2=\frac{6m_1+8m_2}{m_2}=8+6\frac{m_1}{m_2}}.$$

The requested results are therefore

$$\boxed{\displaystyle x_2(t)=\frac{6m_1+8m_2}{m_2}\cos(4t)},$$

$$\boxed{\displaystyle y_2(t)=\frac{6m_1+8m_2}{m_2}\sin(4t)},$$

$$\boxed{\displaystyle R_2=\frac{6m_1+8m_2}{m_2}}.$$

Because all three position vectors contain the same $\cos(4t)$ and $\sin(4t)$ factors, particle 1, particle 2, and the center of mass always lie on the same ray from the origin and rotate together with the same angular speed.

## Choose masses for the visualization

The analytical result works for any positive masses. Change the values below and rerun the remaining cells to explore how the mass ratio changes particle 2's orbit.

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML, display

m1 = 1.0
m2 = 2.0

if m1 <= 0 or m2 <= 0:
    raise ValueError("Both masses must be positive.")

R1 = 2.0
RCM = 8.0
R2 = (6*m1 + 8*m2) / m2
omega = 4.0
period = 2*np.pi / omega

print(f"m1 = {m1:g}, m2 = {m2:g}")
print(f"R1 = {R1:.3f} m")
print(f"RCM = {RCM:.3f} m")
print(f"R2 = {R2:.3f} m")
print(f"Period = {period:.3f} s")
```

## Numerical check of the center-of-mass definition

This cell samples one complete revolution and verifies that the mass-weighted average of the two particle positions reproduces the specified center-of-mass trajectory.

```python
t_check = np.linspace(0, period, 501)
phase = omega*t_check
r1 = R1*np.column_stack((np.cos(phase), np.sin(phase)))
r2 = R2*np.column_stack((np.cos(phase), np.sin(phase)))
r_cm_given = RCM*np.column_stack((np.cos(phase), np.sin(phase)))
r_cm_from_definition = (m1*r1 + m2*r2) / (m1 + m2)
max_error = np.max(np.linalg.norm(r_cm_from_definition - r_cm_given, axis=1))
print(f"Maximum center-of-mass error: {max_error:.3e} m")
assert max_error < 1e-12
```

## Animation

The dashed circles show the three trajectories. The gray segment emphasizes that both particles and their center of mass remain collinear in this particular motion.

```python
n_frames = 160
times = np.linspace(0, period, n_frames, endpoint=False)
angles = np.linspace(0, 2*np.pi, 500)
limit = 1.15*max(R2, RCM)

fig, ax = plt.subplots(figsize=(7, 7))
ax.set_aspect('equal', adjustable='box')
ax.set_xlim(-limit, limit)
ax.set_ylim(-limit, limit)
ax.set_xlabel('x (m)')
ax.set_ylabel('y (m)')
ax.set_title('Two-particle circular motion and center of mass')
ax.grid(alpha=0.22)

ax.plot(R1*np.cos(angles), R1*np.sin(angles), '--', color='tab:blue', alpha=0.45)
ax.plot(RCM*np.cos(angles), RCM*np.sin(angles), '--', color='tab:green', alpha=0.45)
ax.plot(R2*np.cos(angles), R2*np.sin(angles), '--', color='tab:orange', alpha=0.45)
ax.plot(0, 0, 'k+', ms=9, label='origin')

p1, = ax.plot([], [], 'o', color='tab:blue', ms=10, label=r'particle 1 ($m_1$)')
pcm, = ax.plot([], [], 'D', color='tab:green', ms=8, label='center of mass')
p2, = ax.plot([], [], 'o', color='tab:orange', ms=12, label=r'particle 2 ($m_2$)')
ray, = ax.plot([], [], '-', color='0.55', lw=1.5, alpha=0.75)
time_text = ax.text(0.03, 0.96, '', transform=ax.transAxes, va='top', fontsize=11)
ax.legend(loc='lower right')

def update(frame):
    t = times[frame]
    phi = omega*t
    direction = np.array([np.cos(phi), np.sin(phi)])
    q1 = R1*direction
    qcm = RCM*direction
    q2 = R2*direction
    p1.set_data([q1[0]], [q1[1]])
    pcm.set_data([qcm[0]], [qcm[1]])
    p2.set_data([q2[0]], [q2[1]])
    ray.set_data([0, q2[0]], [0, q2[1]])
    time_text.set_text(f't = {t:.2f} s')
    return p1, pcm, p2, ray, time_text

animation = FuncAnimation(fig, update, frames=n_frames, interval=40, blit=True)
plt.close(fig)
display(HTML(animation.to_jshtml()))
```

## Extension: different angular velocities

Now let the particles retain the same radii but rotate with different angular velocities:

$$\mathbf r_1(t)=R_1[\cos(\omega_1t),\sin(\omega_1t)],$$

$$\mathbf r_2(t)=R_2[\cos(\omega_2t+\phi_0),\sin(\omega_2t+\phi_0)].$$

The definition of center of mass is unchanged:

$$\mathbf r_{\mathrm{CM}}(t)=\frac{m_1\mathbf r_1(t)+m_2\mathbf r_2(t)}{m_1+m_2}.$$

However, when $\omega_1\ne\omega_2$, the particles do not remain on the same ray from the origin. Consequently, the center of mass generally does **not** move on the original radius-8 circle—or on any single circle. Its trajectory is the vector sum of two rotating vectors. The green curve in the animation traces that trajectory, while the green marker always remains on the segment joining the two particles.

```python
# Try changing omega1, omega2, and phi0.
omega1 = 4.0             # rad/s
omega2 = 2.5             # rad/s
phi0 = 0.0               # initial phase offset, radians
duration = 4*np.pi        # seconds; closes for omega1/omega2 = 8/5
n_frames_diff = 180
times_diff = np.linspace(0, duration, n_frames_diff)

def positions_different_omegas(t):
    q1 = R1*np.array([np.cos(omega1*t), np.sin(omega1*t)])
    q2 = R2*np.array([np.cos(omega2*t + phi0), np.sin(omega2*t + phi0)])
    qcm = (m1*q1 + m2*q2)/(m1 + m2)
    return q1, q2, qcm

# Precompute the center-of-mass trail.
cm_path = np.array([positions_different_omegas(t)[2] for t in times_diff])
limit_diff = 1.15*max(R1, R2, np.max(np.linalg.norm(cm_path, axis=1)))
angles_diff = np.linspace(0, 2*np.pi, 500)

fig2, ax2 = plt.subplots(figsize=(7, 7))
ax2.set_aspect('equal', adjustable='box')
ax2.set_xlim(-limit_diff, limit_diff)
ax2.set_ylim(-limit_diff, limit_diff)
ax2.set_xlabel('x (m)')
ax2.set_ylabel('y (m)')
ax2.set_title(r'Different angular velocities: $\omega_1 \ne \omega_2$')
ax2.grid(alpha=0.22)
ax2.plot(R1*np.cos(angles_diff), R1*np.sin(angles_diff), '--',
         color='tab:blue', alpha=0.35, label='particle 1 orbit')
ax2.plot(R2*np.cos(angles_diff), R2*np.sin(angles_diff), '--',
         color='tab:orange', alpha=0.35, label='particle 2 orbit')
ax2.plot(0, 0, 'k+', ms=9, label='origin')

p1_diff, = ax2.plot([], [], 'o', color='tab:blue', ms=10, label=r'particle 1 ($m_1$)')
p2_diff, = ax2.plot([], [], 'o', color='tab:orange', ms=12, label=r'particle 2 ($m_2$)')
pcm_diff, = ax2.plot([], [], 'D', color='tab:green', ms=8, label='center of mass')
connector, = ax2.plot([], [], '-', color='0.55', lw=1.5, alpha=0.8)
cm_trail, = ax2.plot([], [], '-', color='tab:green', lw=2, alpha=0.8, label='CM path')
time_text_diff = ax2.text(0.03, 0.96, '', transform=ax2.transAxes, va='top', fontsize=10)
ax2.legend(loc='lower right', fontsize=8)

def update_different_omegas(frame):
    t = times_diff[frame]
    q1, q2, qcm = positions_different_omegas(t)
    p1_diff.set_data([q1[0]], [q1[1]])
    p2_diff.set_data([q2[0]], [q2[1]])
    pcm_diff.set_data([qcm[0]], [qcm[1]])
    connector.set_data([q1[0], q2[0]], [q1[1], q2[1]])
    cm_trail.set_data(cm_path[:frame+1, 0], cm_path[:frame+1, 1])
    time_text_diff.set_text(
        f't = {t:.2f} s\n'
        f'omega1 = {omega1:g} rad/s, omega2 = {omega2:g} rad/s'
    )
    return p1_diff, p2_diff, pcm_diff, connector, cm_trail, time_text_diff

animation_different_omegas = FuncAnimation(
    fig2, update_different_omegas, frames=n_frames_diff, interval=45, blit=True
)
plt.close(fig2)
display(HTML(animation_different_omegas.to_jshtml()))
```
